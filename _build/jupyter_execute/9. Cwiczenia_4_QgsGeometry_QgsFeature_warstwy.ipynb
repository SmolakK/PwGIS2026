## 1. Tworzenie geometrii

Geometrie w PyQGIS mozna tworzyc na trzy sposoby:
z listy wspolrzednych, z formatu WKT lub z formatu WKB.
W tym cwiczeniu skupiamy sie na pierwszych dwoch.

Każdą geometrię dodaj do tymczasowej warstwy w QGIS tak żeby była widoczna na mapie. Poniżej fragment kodu przykładowego do dodania punktu do warstwy tymczasowej. Modyfikuj go odpowiednio.

````python
from qgis.core import QgsVectorLayer, QgsFeature, QgsGeometry, QgsPointXY, QgsProject

wl = QgsVectorLayer('Point?crs=EPSG:4326', 'test', 'memory')

f = QgsFeature()
f.setGeometry(QgsGeometry.fromPointXY(QgsPointXY(17.038, 51.108)))
wl.dataProvider().addFeature(f)

QgsProject.instance().addMapLayer(wl)
iface.mapCanvas().refresh()
````

from qgis.core import (
    QgsVectorFileWriter, QgsProject,
    QgsCoordinateReferenceSystem, QgsCoordinateTransform
)

wl = QgsProject.instance().mapLayersByName('moje_miasta')[0]

opcje = QgsVectorFileWriter.SaveVectorOptions()
opcje.driverName   = 'GPKG'
opcje.fileEncoding = 'UTF-8'

    wl.crs(),

    QgsProject.instance()
)

    wl, 'C:/GIS_kurs/C4/moje_miasta_PL1992.gpkg',
    QgsProject.instance().transformContext(), opcje
)

if blad == QgsVectorFileWriter.NoError:

    # Wczytaj i sprawdz CRS
    from qgis.core import QgsVectorLayer
