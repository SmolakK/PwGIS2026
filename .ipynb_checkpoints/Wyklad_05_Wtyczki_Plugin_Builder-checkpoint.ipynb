## 0. Jak Python organizuje kod — pakiety i moduly

Zanim stworzymy wtyczke, musimy zrozumiec jak Python organizuje
kod w wieksze struktury. Wtyczka QGIS jest pakietem Pythona —
nie zrozumiesz jej struktury bez znajomosci tych podstaw.

### 0.1 Modul

**Modul** to pojedynczy plik `.py`. Kazdy plik Pythona jest modulem.

```
obliczenia.py      <- modul o nazwie 'obliczenia'
wizualizacja.py    <- modul o nazwie 'wizualizacja'
utils.py           <- modul o nazwie 'utils'
```

Importujesz go przez:

```python
import obliczenia
from obliczenia import funkcja_a
from obliczenia import funkcja_a, funkcja_b
```

### 0.2 Pakiet

**Pakiet** to folder zawierajacy moduly i obowiazkowy plik `__init__.py`.
Pakiet pozwala grupowac powiazane moduly w jedna przestrzen nazw.

```
moj_projekt/               <- folder = pakiet
    __init__.py            <- obowiazkowy — bez niego to zwykly folder
    obliczenia.py          <- modul w pakiecie
    wizualizacja.py        <- modul w pakiecie
    utils.py               <- modul w pakiecie
```

Importujesz przez:

```python
import moj_projekt
from moj_projekt import obliczenia
from moj_projekt.obliczenia import funkcja_a
```

### 0.3 Pakiety zagniezdzone

Pakiety moga zawierac inne pakiety — kazdy podfolder
z plikiem `__init__.py` staje sie podpakietem.

```
moj_projekt/
    __init__.py
    obliczenia.py
    ui/                    <- podpakiet
        __init__.py
        dialogi.py
        panele.py
    dane/                  <- podpakiet
        __init__.py
        wczytywanie.py
        zapis.py
```

Importujesz przez:

```python
from moj_projekt.ui import dialogi
from moj_projekt.ui.dialogi import DialogGlowny
from moj_projekt.dane.wczytywanie import wczytaj_gpkg
```

### 0.4 Rola pliku __init__.py

`__init__.py` pelni trzy role:

**Rola 1 — Oznaczenie folderu jako pakietu**

Bez `__init__.py` Python traktuje folder jako zwykly katalog,
nie jako pakiet. Plik moze byc zupelnie pusty — sama jego
obecnosc wystarczy.

```python
# __init__.py moze byc pusty:
# (plik istnieje, ale nie zawiera nic)
```

**Rola 2 — Kod wykonywany przy imporcie pakietu**

Kiedy piszesz `import moj_projekt`, Python wykonuje `__init__.py`.
Mozna tam umiescic inicjalizacje, wspolne importy lub konfiguracje.

```python
# __init__.py — wykonywany przy 'import moj_projekt'

WERSJA = '1.0.0'
AUTOR  = 'Jan Kowalski'

# Importy skrocone — udostepnij najwazniejsze klasy bezposrednio
from .obliczenia import Kalkulator
from .wizualizacja import rysuj_mape
```

Dzieki temu zamiast:
```python
from moj_projekt.obliczenia import Kalkulator
```
mozna pisac krotcej:
```python
from moj_projekt import Kalkulator
```

**Rola 3 — Kontrola publicznego API przez __all__**

```python
# __init__.py
# Definicja co jest dostepne przy 'from moj_projekt import *'
__all__ = ['Kalkulator', 'rysuj_mape']
```

### 0.5 Import wzgledny vs bezwzgledny

Wewnatrz pakietu moduly moga importowac siebie nawzajem
na dwa sposoby:

```python
# Import bezwzgledny — pelna sciezka od korzenia
from moj_projekt.utils import pomocnicza

# Import wzgledny — sciezka wzgledem biezacego modulu
from .utils import pomocnicza    # . = biezacy pakiet
from ..dane import wczytaj       # .. = pakiet nadrzedny
```

W wtyczkach QGIS zawsze uzywamy **importow wzglednych** (z kropka),
bo nie wiemy pod jaka nazwa pakiet zostanie zainstalowany.

```python
# W buffer_tool.py — import wzgledny:
from .buffer_tool_dialog import BufferToolDialog

# NIE tak — import bezwzgledny zadziala tylko u Ciebie:
from buffer_tool.buffer_tool_dialog import BufferToolDialog
```

### 0.6 Biblioteki zewnetrzne — jak sa zbudowane

Popularne biblioteki to nic innego jak pakiety Pythona
zainstalowane w okreslonym miejscu w systemie.

```
geopandas/                 <- pakiet
    __init__.py            <- inicjalizacja, eksportuje GeoDataFrame
    geodataframe.py        <- klasa GeoDataFrame
    geoseries.py           <- klasa GeoSeries
    io/                    <- podpakiet
        __init__.py
        file.py            <- read_file(), to_file()
        arrow.py
    tools/                 <- podpakiet
        __init__.py
        crs.py
        ...
```

Kiedy piszesz `import geopandas as gpd`, Python:

1. Znajduje folder `geopandas/` w sciezce instalacji.
2. Wykonuje `geopandas/__init__.py`.
3. `__init__.py` importuje `GeoDataFrame` z `geodataframe.py`
   i udostepnia go jako `gpd.GeoDataFrame`.

Dlatego mozesz pisac `gpd.GeoDataFrame` zamiast
`gpd.geodataframe.GeoDataFrame` — `__init__.py` to ulatwia.

### 0.7 qgis.core, qgis.gui — jak to dziala

To samo dotyczy PyQGIS:

```python
from qgis.core import QgsVectorLayer
```

Oznacza: w pakiecie `qgis`, w podpakiecie `core`,
znajdz i zaladuj klase `QgsVectorLayer`.

```
qgis/
    __init__.py
    core/
        __init__.py    <- eksportuje QgsVectorLayer, QgsProject, ...
        ...
    gui/
        __init__.py    <- eksportuje QgisInterface, QgsMapCanvas, ...
        ...
    processing/
        __init__.py
        ...
```

Wtyczka QGIS jest dokkladnie taka sama struktura —
pakietem Pythona z `__init__.py` na czele,
tyle ze QGIS wie gdzie go szukac i jak zaladowac.


````python
# Demonstracja struktury pakietu w czystym Pythonie
# Mozna uruchomic w Jupyter — nie wymaga QGIS

import tempfile, os, sys

# Stworz tymczasowy pakiet
tmp = tempfile.mkdtemp()
pkg = os.path.join(tmp, 'moj_pakiet')
os.makedirs(pkg)

# Plik utils.py
with open(os.path.join(pkg, 'utils.py'), 'w') as f:
    f.write('def przywitaj(imie):\n')
    f.write('    return f"Czesc, {imie}!"\n')
    f.write('\n')
    f.write('STALA = 42\n')

# Plik obliczenia.py
with open(os.path.join(pkg, 'obliczenia.py'), 'w') as f:
    f.write('from .utils import STALA\n')
    f.write('\n')
    f.write('def kwadrat(x):\n')
    f.write('    return x ** 2\n')
    f.write('\n')
    f.write('def suma_ze_stala(x):\n')
    f.write('    return x + STALA\n')

# __init__.py — eksportuje najwazniejsze rzeczy
with open(os.path.join(pkg, '__init__.py'), 'w') as f:
    f.write('WERSJA = "1.0.0"\n')
    f.write('\n')
    f.write('# Import skrocony — dostepne bezposrednio z pakietu\n')
    f.write('from .utils import przywitaj\n')
    f.write('from .obliczenia import kwadrat\n')

# Dodaj do sciezki Pythona
sys.path.insert(0, tmp)

# Importuj pakiet
import moj_pakiet

print('--- Import bezposredni z pakietu (przez __init__.py) ---')
print(moj_pakiet.przywitaj('student'))    # dziala przez __init__.py
print(moj_pakiet.kwadrat(5))              # dziala przez __init__.py
print(f'Wersja: {moj_pakiet.WERSJA}')    # zmienna z __init__.py

print()
print('--- Import przez pelna sciezke ---')
from moj_pakiet.obliczenia import suma_ze_stala
print(suma_ze_stala(10))                   # 10 + 42 = 52

print()
print('--- Bez __init__.py Python nie znajdzie pakietu ---')
print('Zawartosci folderu moj_pakiet:')
for f in sorted(os.listdir(pkg)):
    print(f'  {f}')
````

### 0.8 Co musi miec wtyczka QGIS

Wtyczka to pakiet Pythona z dodatkowymi wymaganiami narzuconymi przez QGIS.
Minimalne wymagania:

**1. Folder w katalogu wtyczek**

Folder musi znajdowac sie w:

```
Windows: %APPDATA%\QGIS\QGIS3\profiles\default\python\plugins\
macOS  : ~/Library/Application Support/QGIS/QGIS3/profiles/default/python/plugins/
Linux  : ~/.local/share/QGIS/QGIS3/profiles/default/python/plugins/
```

**2. Plik __init__.py z funkcja classFactory**

```python
# __init__.py — wymagana funkcja classFactory

def classFactory(iface):
    """Wywolywana przez QGIS przy ladowaniu wtyczki.
    Musi zwrocic obiekt glownej klasy wtyczki."""
    from .moja_wtyczka import MojaWtyczka
    return MojaWtyczka(iface)
```

Bez `classFactory` QGIS nie bedzie wiedzial jak uruchomic wtyczke.
Argument `iface` to obiekt `QgisInterface` — dostep do calego interfejsu QGIS.

**3. Plik metadata.txt**

```ini
[general]
name=Nazwa Wtyczki          <- wymagane
qgisMinimumVersion=3.0      <- wymagane
description=Opis wtyczki    <- wymagane
version=0.1                 <- wymagane
author=Autor                <- wymagane
email=email@example.com     <- wymagane
```

Bez poprawnego `metadata.txt` QGIS odrzuci wtyczke przy ladowaniu.

**4. Klasa glowna z trzema metodami**

```python
class MojaWtyczka:

    def __init__(self, iface):
        """Konstruktor — wywolywany raz przy starcie QGIS.
        Przechowuje referencje do interfejsu."""
        self.iface = iface

    def initGui(self):
        """Dodaje elementy do interfejsu QGIS.
        Wywolywana przez QGIS po konstruktorze."""
        # Tu dodajesz przyciski, pozycje menu, panele
        pass

    def unload(self):
        """Usuwa elementy z interfejsu QGIS.
        Wywolywana gdy uzytkownik wylacza wtyczke."""
        # Tu usuwasz wszystko co dodales w initGui
        pass
```

Brak ktorejkolwiek z tych trzech metod spowoduje blad.

### 0.9 Pelne minimalne drzewo plikow

```
moja_wtyczka/              <- nazwa folderu = nazwa modulu
    __init__.py            <- classFactory — obowiazkowy
    moja_wtyczka.py        <- klasa glowna z initGui/unload
    metadata.txt           <- metadane — obowiazkowy
```

Wszystko inne (dialogi, ikony, pliki `.ui`, podpakiety)
jest opcjonalne — dodajesz je w miare jak wtyczka rosnie.

### 0.10 Przepływ sterowania przy starcie QGIS

```
QGIS startuje
    |
    +-- Skanuje folder plugins/
    |   Dla kazdego podfolderu sprawdza metadata.txt
    |
    +-- Jesli wtyczka jest wlaczona w Menedzerze:
    |       import moja_wtyczka         <- wykonuje __init__.py
    |       obj = classFactory(iface)   <- dostaje obiekt wtyczki
    |       obj.initGui()               <- wtyczka dodaje GUI
    |
    QGIS gotowy — ikona wtyczki widoczna w pasku
    |
    Uzytkownik wylacza wtyczke:
        obj.unload()                    <- wtyczka sprzata po sobie
```

## 1. Czym jest wtyczka QGIS

### 1.1 Definicja

Wtyczka (ang. *plugin*) to rozszerzenie QGIS napisane w Pythonie,
ktore dodaje nowe funkcje do aplikacji:

- nowy przycisk lub pozycja w menu,
- okno dialogowe z wlasnym interfejsem,
- nowy algorytm w Processing Toolbox,
- panel dokowany w interfejsie QGIS.

### 1.2 Trzy sposoby programowania w QGIS — przypomnienie

| Sposob | Interfejs | Trwalosc | Dystybucja |
|---|---|---|---|
| Skrypt w konsoli | Brak | Jednorazowy | Plik .py |
| Skrypt w edytorze | Brak | Plik .py | Plik .py |
| **Wtyczka** | **Wlasny GUI** | **Przy kazdym starcie** | **Repozytorium QGIS** |

### 1.3 Jak QGIS laduje wtyczke

```
Start QGIS
    |
    +-- Skanuje folder wtyczek
    |   Windows: C:/Users/<user>/AppData/Roaming/QGIS/QGIS3/profiles/default/python/plugins/
    |   macOS  : ~/Library/Application Support/QGIS/QGIS3/profiles/default/python/plugins/
    |   Linux  : ~/.local/share/QGIS/QGIS3/profiles/default/python/plugins/
    |
    +-- Dla kazdego folderu wczytuje __init__.py
    |
    +-- Wywoluje classFactory(iface) -> zwraca obiekt wtyczki
    |
    +-- Wywoluje plugin.initGui() -> dodaje przyciski i menu
    |
    QGIS gotowy do pracy
```

### 1.4 Cykl zycia wtyczki

```python
# Te metody musi implementowac kazda wtyczka:

def __init__(self, iface):
    # Konstruktor — wywolywany raz przy ladowaniu
    self.iface = iface

def initGui(self):
    # Dodaj przyciski, menu, panele do interfejsu QGIS
    pass

def unload(self):
    # Usun przyciski i menu przy wylaczaniu wtyczki
    pass
```

### 1.5 Minimalna wtyczka

Najprostsza mozliwa wtyczka skladajaca sie z trzech plikow:

```
moja_wtyczka/
    __init__.py       <- wymagany przez Python jako pakiet
    metadata.txt      <- wymagany przez QGIS (nazwa, wersja, autor)
    moja_wtyczka.py   <- glowna klasa wtyczki
```


````python
# Pokaz struktury minimalnej wtyczki — tylko do przejrzenia
# NIE uruchamiaj tego w QGIS

minimal = {
    '__init__.py': '''
def classFactory(iface):
    from .moja_wtyczka import MojaWtyczka
    return MojaWtyczka(iface)
''',
    'metadata.txt': '''
[general]
name=Moja Wtyczka
qgisMinimumVersion=3.0
description=Pierwsza wtyczka
version=0.1
author=Jan Kowalski
email=jan@example.com
''',
    'moja_wtyczka.py': '''
from qgis.PyQt.QtWidgets import QAction
from qgis.PyQt.QtGui import QIcon

class MojaWtyczka:
    def __init__(self, iface):
        self.iface = iface
        self.action = None

    def initGui(self):
        self.action = QAction("Uruchom", self.iface.mainWindow())
        self.action.triggered.connect(self.run)
        self.iface.addToolBarIcon(self.action)

    def unload(self):
        self.iface.removeToolBarIcon(self.action)

    def run(self):
        print("Wtyczka dziala!")
'''
}

print('Struktura minimalnej wtyczki:')
for plik, zawartosc in minimal.items():
    print(f'\n--- {plik} ---')
    print(zawartosc.strip())
````

## 2. Plugin Builder — generowanie szkieletu wtyczki

Pisanie wtyczki od zera jest mozliwe ale zbudne.
**Plugin Builder** generuje kompletny szkielet ze wszystkimi wymaganymi plikami,
poprawna struktura i przykladowym kodem.

### 2.1 Uruchomienie Plugin Builder

`Wtyczki -> Plugin Builder -> Plugin Builder`

Pojawi sie kreator. Wypelnij kolejne ekrany:

### 2.2 Ekran 1 — Podstawowe informacje

| Pole | Przykladowa wartosc | Uwagi |
|---|---|---|
| Class name | `BufferTool` | PascalCase, bez spacji |
| Plugin name | `Buffer Tool` | Nazwa widoczna w QGIS |
| Module name | `buffer_tool` | snake_case, nazwa folderu |
| Plugin directory | *(zostaw domyslny)* | Folder wtyczek QGIS |
| Description | `Narzedzie do buforowania warstw` | Krotki opis |
| Author | Twoje imie i nazwisko | |
| Email | Twoj email | |

### 2.3 Ekran 2 — Szablon

Wybierz: **Tool button with dialog**

To najpopularniejszy typ — przycisk w pasku narzedzi otwierajacy okno dialogowe.

Inne dostepne szablony:

| Szablon | Opis |
|---|---|
| Tool button with dialog | Przycisk + okno dialogowe |
| Tool button with dock widget | Przycisk + panel dokowany |
| Processing provider | Nowy algorytm w Processing Toolbox |

### 2.4 Ekran 3 — Opcje dodatkowe

Zaznacz:

- **Flag the plugin as experimental** — na razie tak, to wersja testowa
- **Add to menu** — tak, chcemy pozycje w menu

Odznacz:

- Internacjonalizacja (i18n) — niepotrzebna na poczatek
- Testy jednostkowe — niepotrzebne na poczatek

### 2.5 Generowanie

Kliknij **Generate**.

Plugin Builder:

1. Tworzy folder z plikami wtyczki w katalogu wtyczek QGIS.
2. Wyswietla podsumowanie z lista plikow i dalszymi krokami.
3. Informuje gdzie jest folder — zanotuj te sciezke!

> Po wygenerowaniu NIE uruchamiaj jeszcze wtyczki.
> Najpierw musimy dostosowac interfejs w Qt Designer.

## 3. Struktura wygenerowanych plikow

Po uruchomieniu Plugin Builder w folderze wtyczek pojawi sie:

```
buffer_tool/
    __init__.py                  <- classFactory — punkt wejscia
    buffer_tool.py               <- glowna klasa wtyczki
    buffer_tool_dialog.py        <- klasa okna dialogowego
    buffer_tool_dialog_base.ui   <- projekt interfejsu (Qt Designer)
    metadata.txt                 <- metadane wtyczki
    resources.py                 <- skompilowane zasoby (ikony)
    resources.qrc                <- definicja zasobow
    icon.png                     <- ikona wtyczki
    pb_tool.cfg                  <- konfiguracja Plugin Builder
    README.html                  <- dokumentacja
```

### 3.1 __init__.py

```python
def classFactory(iface):
    from .buffer_tool import BufferTool
    return BufferTool(iface)
```

Punkt wejscia. QGIS wywoluje `classFactory(iface)` przy starcie
i oczekuje obiektu wtyczki w zamian.

### 3.2 buffer_tool.py — glowna klasa

```python
class BufferTool:

    def __init__(self, iface):
        self.iface = iface        # dostep do interfejsu QGIS
        self.dlg = None           # okno dialogowe
        self.actions = []         # lista akcji (przyciskow)
        self.menu = '&Buffer Tool'

    def initGui(self):
        # Tworzy akcje i dodaje do menu/paska narzedzi
        icon_path = ':/plugins/buffer_tool/icon.png'
        self.add_action(
            icon_path,
            text='Buffer Tool',
            callback=self.run,
            parent=self.iface.mainWindow()
        )

    def unload(self):
        # Usuwa przyciski z interfejsu
        for action in self.actions:
            self.iface.removePluginMenu('&Buffer Tool', action)
            self.iface.removeToolBarIcon(action)

    def run(self):
        # Wywolywane po kliknieciu przycisku
        if self.dlg is None:
            self.dlg = BufferToolDialog()
        self.dlg.show()
        result = self.dlg.exec_()
        if result:
            pass  # TU BEDZIE LOGIKA
```

### 3.3 buffer_tool_dialog_base.ui

Plik XML opisujacy wyglad okna dialogowego.
Edytujemy go **wylacznie w Qt Designer** — nigdy recznie.

```xml
<?xml version="1.0" encoding="UTF-8"?>
<ui version="4.0">
 <class>BufferToolDialogBase</class>
 <widget class="QDialog" name="BufferToolDialogBase">
  <!-- Tu Qt Designer wstawia widgety -->
 </widget>
</ui>
```

### 3.4 buffer_tool_dialog.py

```python
import os
from qgis.PyQt import uic, QtWidgets

FORM_CLASS, _ = uic.loadUiType(os.path.join(
    os.path.dirname(__file__), 'buffer_tool_dialog_base.ui'
))

class BufferToolDialog(QtWidgets.QDialog, FORM_CLASS):
    def __init__(self, parent=None):
        super().__init__(parent)
        self.setupUi(self)  # laduje plik .ui
```

`uic.loadUiType()` czyta plik `.ui` i generuje klase Pythona w locie.
Dzieki temu nie musimy recznie kompilowac interfejsu.

### 3.5 metadata.txt

```ini

version=0.1
```

## 4. Qt Designer — projektowanie interfejsu

Qt Designer to graficzny edytor interfejsow Qt.
Przeciagasz widgety, ustawiasz wlasciwosci, a Designer generuje plik `.ui`.

### 4.1 Uruchomienie Qt Designer

**Windows:**
```
C:\Program Files\QGIS 3.x\bin\designer.exe
```

**macOS:**
```
/Applications/QGIS.app/Contents/MacOS/bin/designer
```

**Linux:**
```bash
designer
# lub
qtdesigner
```

Mozna tez otworzyc plik `.ui` bezposrednio z eksploratora plikow
— kliknij prawym przyciskiem -> Otworz za pomoca -> Qt Designer.

### 4.2 Otworcie pliku .ui wtyczki

1. Uruchom Qt Designer.

### 4.5 QgsMapLayerComboBox — specjalny widget QGIS

`QgsMapLayerComboBox` to widget specyficzny dla QGIS —
automatycznie wypelnia sie warstwami z otwartego projektu.

Zeby go uzyc w Qt Designer:

1. Przeciagnij zwykly `QComboBox` na okno.
2. Kliknij prawym przyciskiem na widget.
3. Wybierz **Promote to...**
4. W polu *Promoted class name* wpisz: `QgsMapLayerComboBox`
5. W polu *Header file* wpisz: `qgis.gui`
6. Kliknij **Add**, potem **Promote**.

### 4.6 Projektowanie dialogu wtyczki — krok po kroku

Zbudujemy okno dialogowe z:

- lista warstw wejsciowych (`QgsMapLayerComboBox`),
- polem na promien bufora (`QDoubleSpinBox`),
- etykietami opisujacymi pola,
- przyciskami OK i Cancel (juz sa — nie usuwaj).

**Krok 1 — Usun domyslna zawartosc**

Usuń wszystkie elementy poza przyciskami OK/Cancel
(jezeli sa jakies domyslne widgety).

**Krok 2 — Dodaj QLabel dla warstwy**

1. Z Widget Box przeciagnij `Label` na okno.
2. Kliknij dwukrotnie i wpisz: `Warstwa wejsciowa:`
3. W Property Editor zmien `objectName` na `lblWarstwa`.

**Krok 3 — Dodaj QgsMapLayerComboBox**

1. Przeciagnij `QComboBox` pod etykiete.
2. Prawym przyciskiem -> `Promote to...`
3. Promoted class: `QgsMapLayerComboBox`, Header: `qgis.gui`
4. Kliknij Add -> Promote.
5. Zmien `objectName` na `mMapLayerComboBox`.

**Krok 4 — Dodaj QLabel dla promienia**

1. Przeciagnij kolejny `Label`.
2. Wpisz: `Promien bufora (km):`
3. `objectName`: `lblPromien`

**Krok 5 — Dodaj QDoubleSpinBox**

1. Przeciagnij `Double Spin Box` pod etykiete.
2. `objectName`: `spbPromien`
3. W Property Editor ustaw:
   - `minimum`: `0.1`
   - `maximum`: `10000.0`
   - `value`: `10.0`
   - `singleStep`: `1.0`

**Krok 6 — Uklad (Layout)**

Zaznacz wszystkie widgety (Ctrl+A), nastepnie:
`Form -> Lay Out Vertically` (lub Ctrl+L)

Layout automatycznie rozmieszcza widgety przy zmianie rozmiaru okna.

**Krok 7 — Zapisz**

`File -> Save` (Ctrl+S)

Plik `.ui` zostaje zaktualizowany.

## 5. Podlaczenie GUI do kodu — sygnaly i sloty

### 5.1 Sygnaly i sloty — mechanizm Qt

Qt komunikuje sie przez **sygnaly** i **sloty**:

- **Sygnal** — zdarzenie emitowane przez widget (klikniecie, zmiana wartosci)
- **Slot** — funkcja wywolywana w odpowiedzi na sygnal

```python
# Skladnia: obiekt.sygnal.connect(slot)
przycisk.clicked.connect(moja_funkcja)
#                ^sygnal  ^slot
```

### 5.2 Najczesciej uzywane sygnaly

| Widget | Sygnal | Kiedy emitowany |
|---|---|---|
| QPushButton | `clicked` | Po kliknieciu przycisku |
| QCheckBox | `stateChanged` | Po zmianie stanu |
| QComboBox | `currentIndexChanged` | Po wybraniu innej pozycji |
| QDoubleSpinBox | `valueChanged` | Po zmianie wartosci |
| QLineEdit | `textChanged` | Po zmianie tekstu |
| QgsMapLayerComboBox | `layerChanged` | Po wybraniu innej warstwy |

### 5.3 Modyfikacja metody run()

Otwieramy plik `buffer_tool.py` i modyfikujemy metode `run()`.
To tutaj bedzie cala logika wtyczki.

```python
def run(self):
    # Stworz dialog jesli jeszcze nie istnieje
    if self.dlg is None:
        self.dlg = BufferToolDialog()

    # Pokaz okno i czekaj na odpowiedz uzytkownika
    self.dlg.show()
    result = self.dlg.exec_()

    # Jesli uzytkownik kliknal OK
    if result:
        self.wykonaj()

def wykonaj(self):
    # Pobierz wartosci z widgetow
    warstwa = self.dlg.mMapLayerComboBox.currentLayer()
    promien = self.dlg.spbPromien.value()

    # Walidacja
    if warstwa is None:
        self.iface.messageBar().pushWarning(
            'Buffer Tool', 'Wybierz warstwe wejsciowa.'
        )
        return

    # Komunikat o sukcesie
    self.iface.messageBar().pushSuccess(
        'Buffer Tool',
        f'Warstwa: {warstwa.name()}, Promien: {promien} km'
    )
```

### 5.4 Pasek komunikatow QGIS

Zamiast `print()` wtyczki uzywaja paska komunikatow QGIS:

```python
bar = self.iface.messageBar()

bar.pushInfo('Tytul',    'Komunikat informacyjny')   # niebieski
bar.pushSuccess('Tytul', 'Operacja zakonczona')      # zielony
bar.pushWarning('Tytul', 'Ostrzezenie')              # zolty
bar.pushCritical('Tytul','Blad krytyczny')           # czerwony
```

### 5.5 Podlaczenie przycisku wyboru pliku

Czesto potrzebujemy przycisku otwierajacego dialog wyboru pliku.
Dodaj `QPushButton` w Qt Designer (`objectName`: `btnWybierzPlik`),
a w kodzie:

```python
# W __init__ dialogu lub w run():
self.dlg.btnWybierzPlik.clicked.connect(self.wybierz_plik)

def wybierz_plik(self):
    from qgis.PyQt.QtWidgets import QFileDialog
    sciezka, _ = QFileDialog.getSaveFileName(
        self.dlg,
        'Zapisz jako',
        '',
        'GeoPackage (*.gpkg)'
    )
    if sciezka:
        self.dlg.ledSciezka.setText(sciezka)
```

## 6. Uruchomienie i testowanie wtyczki

### 6.1 Pierwsze uruchomienie

1. W QGIS przejdz do `Wtyczki -> Zarządzaj wtyczkami`.
2. W zakladce **Zainstalowane** znajdz swoja wtyczke (`Buffer Tool`).
3. Jesli nie widac — sprawdz **Experimental** w ustawieniach menedzera.
4. Zaznacz checkbox obok nazwy wtyczki.
5. Kliknij **Zamknij**.

W pasku narzedzi QGIS pojawi sie ikona wtyczki.

### 6.2 Jesli wtyczka sie nie laduje

Sprawdz w konsoli Pythona:

```python
# Wyswietl ostatni blad ladowania wtyczek
import traceback
from qgis.utils import plugin_builder_errors

# Lub sprawdz logi QGIS:
# Menu: Widok -> Panele -> Log Messages
```

Najczestsze bledy przy pierwszym uruchomieniu:

| Blad | Przyczyna | Rozwiazanie |
|---|---|---|
| `ImportError` | Brak modulu | Sprawdz importy w .py |
| `SyntaxError` | Blad skladni | Sprawdz wciety kod |
| `AttributeError` | Zly objectName widgetu | Sprawdz nazwy w Qt Designer |
| Ikona nie pojawia sie | Brak resources.py | Uruchom `pyrcc5` |

### 6.3 Plugin Reloader — przeladowanie bez restartu

Za kazdym razem gdy zmieniasz kod wtyczki, normalnie trzeba
zrestartowac QGIS. **Plugin Reloader** eliminuje ten problem.

1. Zainstaluj Plugin Reloader przez Menedzera wtyczek.
2. `Wtyczki -> Plugin Reloader -> Configure` — wybierz swoja wtyczke.
3. Po kazdej zmianie kodu nacisnij `Ctrl+F5` — wtyczka zostanie przeladowana.

> Plugin Reloader to jedno z najwazniejszych narzedzi przy tworzeniu wtyczek.
> Bez niego kazda poprawka kodu wymaga restartu QGIS (~30 sekund).
> Z nim — przeladowanie trwa sekunde.

### 6.4 Sprawdzenie dzialania

Po uruchomieniu wtyczki:

1. Kliknij ikone wtyczki — powinno otworzyc sie okno dialogowe.
2. Sprawdz czy `QgsMapLayerComboBox` pokazuje warstwy z projektu.
3. Zmien promien w `QDoubleSpinBox`.
4. Kliknij OK — w pasku komunikatow powinien pojawic sie komunikat.

### 6.5 Typowy workflow tworzenia wtyczki

```
1. Zmien kod w buffer_tool.py
       |
2. Zapisz plik (Ctrl+S)
       |
3. Przeladuj wtyczke (Ctrl+F5 przez Plugin Reloader)
       |
4. Przetestuj w QGIS
       |
5. Jesli blad — sprawdz Log Messages lub konsole Pythona
       |
6. Wróc do kroku 1
```

### 6.6 Folder wtyczek — uzyteczne komendy w konsoli

```python
# Znajdz sciezke do folderu wtyczek
import qgis.utils
print(qgis.utils.plugin_paths)

# Sprawdz ktore wtyczki sa zaladowane
print(list(qgis.utils.plugins.keys()))

# Przeladuj wtyczke recznie z konsoli
qgis.utils.reloadPlugin('buffer_tool')
```


````python
# Kompletny kod buffer_tool.py po modyfikacjach z tego wykladu
# Wklej do pliku buffer_tool.py w folderze wtyczki

KOD_GLOWNY = '''
# buffer_tool.py

import os
from qgis.PyQt.QtWidgets import QAction
from qgis.PyQt.QtGui import QIcon
from qgis.core import QgsProject
from .buffer_tool_dialog import BufferToolDialog


class BufferTool:

    def __init__(self, iface):
        self.iface   = iface
        self.dlg     = None
        self.actions = []
        self.menu    = "&Buffer Tool"

    def add_action(self, icon_path, text, callback,
                   enabled_flag=True, add_to_menu=True,
                   add_to_toolbar=True, parent=None):
        icon   = QIcon(icon_path)
        action = QAction(icon, text, parent)
        action.triggered.connect(callback)
        action.setEnabled(enabled_flag)
        if add_to_toolbar:
            self.iface.addToolBarIcon(action)
        if add_to_menu:
            self.iface.addPluginToMenu(self.menu, action)
        self.actions.append(action)
        return action

    def initGui(self):
        icon_path = os.path.join(os.path.dirname(__file__), 'icon.png')
        self.add_action(
            icon_path,
            text='Buffer Tool',
            callback=self.run,
            parent=self.iface.mainWindow()
        )

    def unload(self):
        for action in self.actions:
            self.iface.removePluginMenu(self.menu, action)
            self.iface.removeToolBarIcon(action)

    def run(self):
        if self.dlg is None:
            self.dlg = BufferToolDialog()
        self.dlg.show()
        result = self.dlg.exec_()
        if result:
            self.wykonaj()

    def wykonaj(self):
        warstwa = self.dlg.mMapLayerComboBox.currentLayer()
        promien  = self.dlg.spbPromien.value()

        if warstwa is None:
            self.iface.messageBar().pushWarning(
                "Buffer Tool", "Wybierz warstwe wejsciowa."
            )
            return

        self.iface.messageBar().pushInfo(
            "Buffer Tool",
            f"Warstwa: {warstwa.name()}, Promien: {promien} km — "
            f"gotowe do implementacji"
        )
'''

print('Kod do wklejenia do buffer_tool.py:')
print(KOD_GLOWNY)
````
