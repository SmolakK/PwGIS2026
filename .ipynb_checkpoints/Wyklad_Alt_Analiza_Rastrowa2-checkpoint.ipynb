### Czym rozni sie raster od wektora — przypomnienie

Dane wektorowe opisuja swiat jako **obiekty** (punkty, linie, poligony).
Dane rastrowe opisuja swiat jako **siatke komorek** (pikseli),
gdzie kazda komorka ma jedna lub wiecej wartosci liczbowych.

```
Wektor:                     Raster:
                            +---+---+---+---+
    /\                      |231|245|198|187|
   /  \                     +---+---+---+---+
  / las\                    |220|234|201|192|
 /______\                   +---+---+---+---+
                            |198|210|245|230|
Jeden obiekt = jeden polygon +---+---+---+---+
                            Jeden piksel = jedna wartosc
```

### Kiedy uzywac danych rastrowych

| Zjawisko | Dlaczego raster |
|---|---|
| Numeryczny Model Terenu (NMT) | Wysokosc jest ciagla — kazdemu punktowi odpowiada jedna wartosc |
| Zobrazowanie satelitarne | Kamera rejestruje swiato w pikselach |
| Temperatura powietrza | Zjawisko cigle, interpolowane na siatke |
| Pokrycie terenu | Klasyfikacja na regularnej siatce |
| Opad atmosferyczny | Mierzone w punktach, interpolowane jako raster |

## 1. Model danych rastrowych

### 1.1 Podstawowe pojecia

**Piksel** — podstawowa jednostka rastra. Kazdy piksel:

- zajmuje okreslony obszar na ziemi (rozdzielczosc przestrzenna),
- ma jedna lub wiecej wartosci numerycznych (pasma),
- jest prostokątem (najczesciej kwadratem).

**Rozdzielczosc przestrzenna** — rozmiar piksela na ziemi.
Raster 10 m ma piksele pokrywajace 10 m × 10 m.
Im mniejszy piksel, tym wiecej szczegolów — i wiekszy plik.

| Satelita | Rozdzielczosc | Zastosowanie |
|---|---|---|
| Sentinel-2 | 10 m / 20 m / 60 m | Monitoring, rolnictwo |
| Landsat 8/9 | 30 m | Dlugoletni monitoring |
| SPOT-7 | 1.5 m | Szczegolowe mapy |
| Drone ortofoto | 5–20 cm | Precyzja rolnicza, budowlana |

**Pasmo (band)** — jedna warstwa wartosci. Zobrazowanie RGB ma 3 pasma.
Satelita Sentinel-2 ma 13 pasm (rozny zakres spektralny).

### 1.2 Wspolrzedne rastra — geotransformacja

Raster wie gdzie jest na Ziemi przez **geotransformacje** — szesc liczb:

```
transform = (x_lewy_gory, rozdzielczosc_x, 0,
             y_gory,       0,               -rozdzielczosc_y)
```

Czyli:

- `x_lewy_gory` — wspolrzedna X lewego gornego rogu
- `y_gory` — wspolrzedna Y lewego gornego rogu
- `rozdzielczosc_x` — rozmiar piksela w kierunku X (zazwyczaj dodatni)
- `-rozdzielczosc_y` — rozmiar piksela w kierunku Y (ujemny — Y maleje w dol)

Dzieki geotransformacji mozna przeliczac miedzy numerem wiersza/kolumny
a rzeczywistymi wspolrzednymi geograficznymi.

### 1.3 Wartosc NoData

Nie kazdy piksel ma sensowna wartosc — poza obszarem pomiaru
lub tam gdzie dane sa niedostepne, uzywana jest wartosc **NoData**.
Typowe wartosci: `-9999`, `-32768`, `255`, `nan`.

Kluczowe: NoData **nie jest zerem**. Nalezy ja traktowac jak brak danych,
nie jak wartosc. Wiele bledow analitycznych pochodzi z ignorowania NoData.

### 1.4 Popularne formaty rastrowe

| Format | Rozszerzenie | Uwagi |
|---|---|---|
| GeoTIFF | `.tif`, `.tiff` | Standard — geotransformacja wbudowana w plik |
| Cloud Optimized GeoTIFF | `.tif` (COG) | Zoptymalizowany do streamingu z serwera |
| NetCDF | `.nc` | Dane naukowe, czas jako wymiar |
| HDF5 | `.h5`, `.hdf` | Duze dane satelitarne |
| ASCII Grid | `.asc` | Prosty format tekstowy, przestarzaly |
| JPEG2000 | `.jp2` | Kompresja stratna, Sentinel-2 |

Na tym wykladzie uzywamy **GeoTIFF** — jest wszedobylski i latwy w obsludze.

## 2. Wczytywanie rastra w PyQGIS — QgsRasterLayer

### 2.1 Klasa QgsRasterLayer

`QgsRasterLayer` to odpowiednik `QgsVectorLayer` dla danych rastrowych.
Dziedziczy po `QgsMapLayer` i dodaje metody specyficzne dla rastrow.

### 2.2 Wczytanie pliku

```python
# [QGIS] Wczytaj raster z pliku
from qgis.core import QgsRasterLayer, QgsProject

sciezka = 'C:/GIS_kurs/nmt.tif'   # zmien na swoja sciezke

raster = QgsRasterLayer(sciezka, 'NMT', 'gdal')
#                        ^sciezka  ^nazwa  ^dostawca

if raster.isValid():
    print('Raster wczytany poprawnie!')
    QgsProject.instance().addMapLayer(raster)
else:
    print(f'Blad wczytywania: {raster.error().message()}')
```

### 2.3 Podstawowe metadane

```python
# [QGIS] Metadane rastra
raster = iface.activeLayer()  # zaznacz warstwe rastrowa w Panelu warstw

print(f'Nazwa          : {raster.name()}')
print(f'CRS            : {raster.crs().authid()}')
print(f'Liczba pasm    : {raster.bandCount()}')
print(f'Szerokosc (px) : {raster.width()}')
print(f'Wysokosc  (px) : {raster.height()}')

# Rozdzielczosc
print(f'Rozdzielczosc X: {raster.rasterUnitsPerPixelX():.4f}')
print(f'Rozdzielczosc Y: {raster.rasterUnitsPerPixelY():.4f}')

# Zasieg
ext = raster.extent()
print(f'xMin: {ext.xMinimum():.4f}')
print(f'xMax: {ext.xMaximum():.4f}')
print(f'yMin: {ext.yMinimum():.4f}')
print(f'yMax: {ext.yMaximum():.4f}')
```

### 2.4 Statystyki pasma

```python
# [QGIS] Statystyki wartosci pikseli
from qgis.core import QgsRasterBandStats

raster = iface.activeLayer()
numer_pasma = 1  # pasma numerowane od 1

stats = raster.dataProvider().bandStatistics(
    numer_pasma,
    QgsRasterBandStats.All
)

print(f'Minimum  : {stats.minimumValue:.2f}')
print(f'Maksimum : {stats.maximumValue:.2f}')
print(f'Srednia  : {stats.mean:.2f}')
print(f'Odch.std : {stats.stdDev:.2f}')
```

### 2.5 Odczyt wartosci w punkcie

```python
# [QGIS] Odczytaj wartosc piksela w danym punkcie
from qgis.core import QgsPointXY

raster = iface.activeLayer()

# Wspolrzedne punktu (musza byc w CRS rastra)
punkt = QgsPointXY(17.038, 51.108)   # Wroclaw

wynik, ok = raster.dataProvider().sample(punkt, 1)  # 1 = numer pasma

if ok:
    print(f'Wartosc w punkcie: {wynik:.2f}')
else:
    print('Punkt poza zasiegiem rastra lub wartosc NoData')
```

### 2.6 Informacje o kazdym pasmie

```python
# [QGIS] Przejdz przez wszystkie pasma
raster = iface.activeLayer()

print(f'Raster ma {raster.bandCount()} pasm:')
for i in range(1, raster.bandCount() + 1):
    nazwa = raster.dataProvider().generateBandName(i)
    typ   = raster.dataProvider().dataType(i)
    print(f'  Pasmo {i}: {nazwa} (typ danych: {typ})')
```


````python
# [JUPYTER] Demonstracja metadanych rastra przez rasterio
# Identyczne informacje co w PyQGIS — inny interfejs

try:
    import rasterio
    import numpy as np

    # Stworz przykladowy raster w pamieci do demonstracji
    from rasterio.io import MemoryFile
    from rasterio.transform import from_bounds

    # Sztuczne dane: NMT Polski 50x50 pikseli
    transform = from_bounds(14.1, 49.0, 24.2, 55.0, 50, 50)
    data = np.random.randint(100, 1500, (50, 50), dtype=np.int16)

    with MemoryFile() as memfile:
        with memfile.open(
            driver='GTiff', height=50, width=50,
            count=1, dtype='int16', crs='EPSG:4326',
            transform=transform, nodata=-9999
        ) as dst:
            dst.write(data, 1)

        with memfile.open() as src:
            print('=== Metadane rastra (przez rasterio) ===')
            print(f'Sterownik      : {src.driver}')
            print(f'CRS            : {src.crs}')
            print(f'Liczba pasm    : {src.count}')
            print(f'Szerokosc (px) : {src.width}')
            print(f'Wysokosc  (px) : {src.height}')
            print(f'Typ danych     : {src.dtypes}')
            print(f'NoData         : {src.nodata}')
            print(f'Geotransformacja: {src.transform}')
            print(f'Zasieg         : {src.bounds}')
            print()

            arr = src.read(1)
            print(f'Ksztalt tablicy NumPy: {arr.shape}')
            print(f'Min wartosci   : {arr.min()}')
            print(f'Max wartosci   : {arr.max()}')
            print(f'Srednia        : {arr.mean():.2f}')

except ImportError:
    print('Zainstaluj: pip install rasterio numpy')
````

## 3. Wczytywanie rastra w rasterio

### 3.1 Czym jest rasterio

`rasterio` to biblioteka Pythona do odczytu i zapisu danych rastrowych.
Dziala niezaleznie od QGIS, na bazie biblioteki GDAL.
Zwraca dane jako tablice NumPy — mozna na nich wykonywac dowolne obliczenia.

### 3.2 Podstawowy odczyt

```python
import rasterio
import numpy as np

# Otwarcie pliku (jak open() dla zwyklych plikow)
with rasterio.open('nmt.tif') as src:

    # Metadane
    print(src.crs)          # uklad wspolrzednych
    print(src.count)        # liczba pasm
    print(src.width)        # szerokosc w pikselach
    print(src.height)       # wysokosc w pikselach
    print(src.dtypes)       # typy danych per pasmo
    print(src.nodata)       # wartosc NoData
    print(src.transform)    # geotransformacja
    print(src.bounds)       # zasieg (left, bottom, right, top)

    # Odczyt danych jako tablica NumPy
    pasmo1 = src.read(1)        # jedno pasmo, shape: (height, width)
    wszystkie = src.read()      # wszystkie pasma, shape: (count, height, width)

print(type(pasmo1))             # numpy.ndarray
print(pasmo1.shape)             # (height, width)
print(pasmo1.dtype)             # int16, float32, uint8...
```

### 3.3 Maska NoData

```python
with rasterio.open('nmt.tif') as src:
    dane = src.read(1)              # surowe dane z NoData
    maska = src.dataset_mask()      # 0 = NoData, 255 = dane

    # Zamien NoData na NaN (dla float)
    dane_float = dane.astype(float)
    dane_float[maska == 0] = np.nan

    # Obliczenia tylko na poprawnych pikselach
    print(f'Srednia (z NaN): {np.nanmean(dane_float):.2f}')
    print(f'Pikselei NoData: {np.sum(maska == 0)}')
```

### 3.4 Przeliczanie wspolrzednych

```python
from rasterio.transform import rowcol, xy

with rasterio.open('nmt.tif') as src:
    # Wspolrzedne geograficzne -> numer wiersza i kolumny
    row, col = rowcol(src.transform, 17.038, 51.108)
    print(f'Wroclaw: wiersz={row}, kolumna={col}')

    # Wartosc w tym pikselu
    dane = src.read(1)
    print(f'Wartosc NMT w Wroclawiu: {dane[row, col]}')

    # Numer wiersza/kolumny -> wspolrzedne centrum piksela
    lon, lat = xy(src.transform, row, col)
    print(f'Centrum piksela: lon={lon:.4f}, lat={lat:.4f}')
```


````python
# [JUPYTER] Demonstracja rasterio z syntetycznym NMT

try:
    import numpy as np
    import matplotlib.pyplot as plt
    import matplotlib.colors as mcolors
    from rasterio.io import MemoryFile
    from rasterio.transform import from_bounds, rowcol, xy

    # Stworz syntetyczny NMT — gradient wysokosci
    W, H = 100, 80
    x = np.linspace(0, 1, W)
    y = np.linspace(0, 1, H)
    xx, yy = np.meshgrid(x, y)

    # Syntetyczna rzezba terenu
    nmt = (
        500 + 800 * np.exp(-((xx-0.3)**2 + (yy-0.4)**2) / 0.05)
            + 400 * np.exp(-((xx-0.7)**2 + (yy-0.6)**2) / 0.03)
            + 50  * np.random.randn(H, W)
    ).astype(np.float32)

    transform = from_bounds(14.1, 49.0, 24.2, 55.0, W, H)

    with MemoryFile() as mf:
        with mf.open(driver='GTiff', height=H, width=W,
                     count=1, dtype='float32', crs='EPSG:4326',
                     transform=transform, nodata=-9999.0) as dst:
            dst.write(nmt, 1)

        with mf.open() as src:
            dane = src.read(1)

            print(f'Ksztalt: {dane.shape}')
            print(f'Min: {dane.min():.1f} m')
            print(f'Max: {dane.max():.1f} m')
            print(f'Srednia: {dane.mean():.1f} m')

            # Wizualizacja
            fig, axes = plt.subplots(1, 2, figsize=(13, 5))

            im0 = axes[0].imshow(dane, cmap='terrain', aspect='auto')
            plt.colorbar(im0, ax=axes[0], label='Wysokosc (m)')
            axes[0].set_title('NMT — paleta terrain')
            axes[0].set_xlabel('Kolumna (piksel)')
            axes[0].set_ylabel('Wiersz (piksel)')

            im1 = axes[1].imshow(dane, cmap='viridis', aspect='auto')
            plt.colorbar(im1, ax=axes[1], label='Wysokosc (m)')
            axes[1].set_title('NMT — paleta viridis')

            plt.suptitle('Syntetyczny Numeryczny Model Terenu', fontsize=13)
            plt.tight_layout()
            plt.show()

except ImportError:
    print('Zainstaluj: pip install rasterio numpy matplotlib')
````

## 4. Operacje na pasmach i pikselach — NumPy

### 4.1 Dlaczego NumPy

Raster wczytany przez rasterio to tablica NumPy.
NumPy pozwala wykonywac operacje na milionach pikseli naraz —
znacznie szybciej niz petla po kazdym pikselu.

Przyklad: raster 1000×1000 pikseli ma 1 000 000 komorek.
Petla Python: kilka sekund. Operacja NumPy: ulamek sekundy.

### 4.2 Podstawowe operacje arytmetyczne

```python
import numpy as np

# dane to tablica NumPy shape (height, width)
# Kazda operacja dziala na kazdym pikselu jednoczesnie

# Przeliczenie metrow na stopy
dane_stopy = dane * 3.28084

# Normalizacja do zakresu 0-1
dane_norm = (dane - dane.min()) / (dane.max() - dane.min())

# Dodanie stalej do kazdego piksela
dane_przesuniete = dane + 100

# Pierwiastek kwadratowy z kazdego piksela
dane_sqrt = np.sqrt(np.maximum(dane, 0))  # maximum by uniknac ujemnych
```

### 4.3 Operacje miedzy pasmami — NDVI

NDVI (Normalized Difference Vegetation Index) to klasyczny wskaznik
zdrowia roslinnosci obliczany z danych wielospektralnych:

```
NDVI = (NIR - RED) / (NIR + RED)
```

gdzie NIR = pasmo bliskiej podczerwieni, RED = pasmo czerwone.
NDVI przyjmuje wartosci od -1 do 1:

| Wartosc | Interpretacja |
|---|---|
| > 0.6 | Gesta roslinnosc (las) |
| 0.2 – 0.6 | Roslinnosc rozrzedzona (trawy, uprawy) |
| 0.0 – 0.2 | Slaba roslinnosc lub gleba |
| < 0.0 | Woda, snieg, skaly |

```python
with rasterio.open('sentinel2.tif') as src:
    # Sentinel-2: pasmo 4 = RED, pasmo 8 = NIR
    red = src.read(4).astype(float)
    nir = src.read(8).astype(float)

# Unikamy dzielenia przez zero
mianownik = nir + red
ndvi = np.where(
    mianownik != 0,
    (nir - red) / mianownik,
    np.nan
)

print(f'NDVI min: {np.nanmin(ndvi):.3f}')
print(f'NDVI max: {np.nanmax(ndvi):.3f}')
print(f'NDVI mean: {np.nanmean(ndvi):.3f}')
```

### 4.4 Filtrowanie pikseli — maski logiczne

```python
# Znajdz piksele wyzsze niz 500 m
maska_wysokie = dane > 500
print(f'Pikseli > 500m: {maska_wysokie.sum()}')

# Oblicz srednia tylko dla wysokich terenow
srednia_wysokie = dane[maska_wysokie].mean()
print(f'Srednia wysokosc terenow > 500m: {srednia_wysokie:.1f}')

# Pokoloruj piksele NoData na zero
dane_clean = dane.copy()
dane_clean[dane == -9999] = 0

# Znajdz piksele w zakresie 200-500 m
niziny = (dane >= 0) & (dane <= 200)
print(f'Procent nizin: {niziny.mean() * 100:.1f}%')
```


````python
# [JUPYTER] Operacje NumPy na rastrze — demonstracja

try:
    import numpy as np
    import matplotlib.pyplot as plt

    # Syntetyczny NMT — Polska
    np.random.seed(42)
    H, W = 80, 100
    x = np.linspace(0, 1, W)
    y = np.linspace(0, 1, H)
    xx, yy = np.meshgrid(x, y)
    nmt = (200 + 1200 * np.exp(-((xx-0.35)**2 + (yy-0.45)**2)/0.06)
               + 600 * np.exp(-((xx-0.70)**2 + (yy-0.55)**2)/0.04)
               + 30 * np.random.randn(H, W)).astype(float)
    nmt = np.clip(nmt, 100, 1600)

    # Stworz syntetyczne pasma jak Sentinel-2
    red = np.clip(0.3 - 0.2 * (nmt / 1600) + 0.05 * np.random.randn(H, W), 0, 1)
    nir = np.clip(0.5 + 0.3 * np.exp(-((xx-0.35)**2+(yy-0.45)**2)/0.08)
                      + 0.05 * np.random.randn(H, W), 0, 1)

    # Oblicz NDVI
    mianownik = nir + red
    ndvi = np.where(mianownik > 0, (nir - red) / mianownik, np.nan)

    # Maski logiczne
    niziny   = nmt < 300
    wzgorza  = (nmt >= 300) & (nmt < 700)
    gory     = nmt >= 700

    print('=== Analiza NMT ===')
    print(f'Niziny   (< 300m) : {niziny.sum():5d} pikseli ({niziny.mean()*100:.1f}%)')
    print(f'Wzgorza (300-700m): {wzgorza.sum():5d} pikseli ({wzgorza.mean()*100:.1f}%)')
    print(f'Gory     (> 700m) : {gory.sum():5d} pikseli ({gory.mean()*100:.1f}%)')
    print()
    print('=== Analiza NDVI ===')
    print(f'NDVI min  : {np.nanmin(ndvi):.3f}')
    print(f'NDVI max  : {np.nanmax(ndvi):.3f}')
    print(f'NDVI mean : {np.nanmean(ndvi):.3f}')

    # Wizualizacja
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))

    im0 = axes[0].imshow(nmt, cmap='terrain', aspect='auto')
    plt.colorbar(im0, ax=axes[0], label='m n.p.m.')
    axes[0].set_title('NMT')

    im1 = axes[1].imshow(ndvi, cmap='RdYlGn', vmin=-0.3, vmax=0.8, aspect='auto')
    plt.colorbar(im1, ax=axes[1], label='NDVI')
    axes[1].set_title('NDVI')

    # Mapa stref wysokosciowych
    strefy = np.zeros_like(nmt, dtype=int)
    strefy[wzgorza] = 1
    strefy[gory]    = 2
    im2 = axes[2].imshow(strefy, cmap='Set1', aspect='auto', vmin=0, vmax=2)
    cbar = plt.colorbar(im2, ax=axes[2], ticks=[0,1,2])
    cbar.ax.set_yticklabels(['Niziny', 'Wzgorza', 'Gory'])
    axes[2].set_title('Strefy wysokosciowe')

    plt.suptitle('Operacje NumPy na danych rastrowych', fontsize=13)
    plt.tight_layout()
    plt.show()

except ImportError:
    print('Zainstaluj: pip install numpy matplotlib')
````

### Jak wykonac te same operacje w QGIS

#### Kalkulator rastrowy — interfejs graficzny

Wszystkie operacje arytmetyczne na pasmach mozna wykonac przez:
`Raster -> Kalkulator rastrowy`

Skladnia nazwy pasma w kalkulatorze:

```
"nazwa_warstwy@numer_pasma"
```

Nazwy dostepnych warstw i pasm widac w panelu po lewej stronie —
mozna je klikac zamiast wpisywac recznie.

**NDVI:**
```
("sentinel2@4" - "sentinel2@3") / ("sentinel2@4" + "sentinel2@3")
```

**Maska logiczna** — piksele w zakresie 200–500 m:
```
("nmt@1" >= 200) AND ("nmt@1" < 500)
```
Wynik: raster z wartosciami 0 (falsz) i 1 (prawda).

**Przeliczenie metrow na stopy:**
```
"nmt@1" * 3.28084
```

**Normalizacja 0–1** (gdy znasz min i max z wlasciwosci warstwy):
```
("nmt@1" - 80) / (280 - 80)
```

#### Statystyki pasma — przez wlasciwosci warstwy

1. Kliknij dwukrotnie warstwe rastrowa w Panelu warstw
2. Zakladka **Histogram** -> kliknij `Oblicz histogram`
3. Zakladka **Informacje** -> sekcja `Statystyki pasma`

Lub przez Processing:
`Przetwarzanie -> Skrzynka narzedzi -> QGIS -> Statystyki rastra`


````python
# [QGIS] Operacje na pasmach przez PyQGIS
# Wklej do Konsoli Pythona QGIS

import processing
from qgis.core import QgsProject

# Pobierz aktywna warstwe rastrowa
raster = iface.activeLayer()
nazwa  = raster.name()

# --- NDVI przez Kalkulator rastrowy Processing ---
# Zakladamy ze pasmo 4 = NIR, pasmo 3 = RED
ndvi = processing.run('qgis:rastercalculator', {
    'EXPRESSION': f'("{nazwa}@4" - "{nazwa}@3") / ("{nazwa}@4" + "{nazwa}@3")',
    'LAYERS'    : [raster],
    'CELLSIZE'  : 0,
    'EXTENT'    : None,
    'CRS'       : raster.crs(),
    'OUTPUT'    : 'TEMPORARY_OUTPUT'
})
ndvi_layer = ndvi['OUTPUT']
ndvi_layer.setName('NDVI')
QgsProject.instance().addMapLayer(ndvi_layer)
print('NDVI obliczony i dodany do projektu.')

# --- Maska logiczna: piksele powyzej 500m ---
# (zakladamy ze pasmo 1 to NMT w metrach)
maska = processing.run('qgis:rastercalculator', {
    'EXPRESSION': f'"{nazwa}@1" > 500',
    'LAYERS'    : [raster],
    'CELLSIZE'  : 0,
    'EXTENT'    : None,
    'CRS'       : raster.crs(),
    'OUTPUT'    : 'TEMPORARY_OUTPUT'
})
maska_layer = maska['OUTPUT']
maska_layer.setName('Maska_powyzej_500m')
QgsProject.instance().addMapLayer(maska_layer)
print('Maska logiczna dodana do projektu.')

# --- Statystyki pasma przez PyQGIS ---
from qgis.core import QgsRasterBandStats

print(f'\nStatystyki pasm warstwy: {raster.name()}')
print(f'{"Pasmo":<8} {"Min":>10} {"Max":>10} {"Srednia":>10} {"Std":>10}')
print('-' * 52)
for i in range(1, raster.bandCount() + 1):
    stats = raster.dataProvider().bandStatistics(
        i, QgsRasterBandStats.All
    )
    print(f'{i:<8} {stats.minimumValue:>10.3f} {stats.maximumValue:>10.3f}'
          f' {stats.mean:>10.3f} {stats.stdDev:>10.3f}')

# --- Przeliczenie metrow na stopy (pasmo 1) ---
stopy = processing.run('qgis:rastercalculator', {
    'EXPRESSION': f'"{nazwa}@1" * 3.28084',
    'LAYERS'    : [raster],
    'CELLSIZE'  : 0,
    'EXTENT'    : None,
    'CRS'       : raster.crs(),
    'OUTPUT'    : 'TEMPORARY_OUTPUT'
})
stopy_layer = stopy['OUTPUT']
stopy_layer.setName('NMT_stopy')
QgsProject.instance().addMapLayer(stopy_layer)
print('\nPasmo 1 przeliczone na stopy i dodane do projektu.')

iface.mapCanvas().refresh()
````

## 5. Przycinanie rastra do obszaru — clip by mask

### 5.1 Po co przycinamy

Dane rastrowe czesto pokrywaja wiekszy obszar niz potrzebujemy.
Satelitarne sceny pokrywaja setki km². Jezeli interesuje nas
tylko jeden powiat, przycinamy raster do jego granicy.

Efekty przycinania:

- mniejszy plik — szybszy odczyt i obliczenia,
- wyniki analizy ograniczone do obszaru zainteresowania,
- piksele poza obszarem maskowane jako NoData.

### 5.2 Przycinanie w PyQGIS przez Processing

```python
# [QGIS] Przytnij raster do granicy warstwy wektorowej
import processing

raster  = QgsProject.instance().mapLayersByName('NMT')[0]
maska   = QgsProject.instance().mapLayersByName('granica_obszaru')[0]

wynik = processing.run('gdal:cliprasterbymasklayer', {
    'INPUT'   : raster,
    'MASK'    : maska,
    'NODATA'  : -9999,
    'CROP_TO_CUTLINE': True,    # przytnij do bounding box maski
    'KEEP_RESOLUTION': True,    # zachowaj oryginalna rozdzielczosc
    'OUTPUT'  : 'TEMPORARY_OUTPUT'
})

raster_przyciety = wynik['OUTPUT']
print(f'Oryginalny:  {raster.width()} x {raster.height()} px')
print(f'Po przyciciu: {raster_przyciety.width()} x {raster_przyciety.height()} px')

# Dodaj do projektu
raster_przyciety.setName('NMT_obszar')
QgsProject.instance().addMapLayer(raster_przyciety)
```

### 5.4 Przycinanie w rasterio

```python
import rasterio
import geopandas as gpd
from rasterio.mask import mask as rasterio_mask
from shapely.geometry import mapping

# Wczytaj mase (granice obszaru)
obszar = gpd.read_file('granica.gpkg')

with rasterio.open('nmt.tif') as src:
    # Upewnij sie ze obszar jest w tym samym CRS co raster
    obszar_repr = obszar.to_crs(src.crs)

    # Lista geometrii jako GeoJSON
    geometrie = [mapping(geom) for geom in obszar_repr.geometry]

    # Przycinanie
    dane_przyciete, transform_nowy = rasterio_mask(
        src, geometrie,
        crop=True,          # przytnij do zasięgu
        nodata=-9999
    )
    meta_nowe = src.meta.copy()
    meta_nowe.update({
        'height': dane_przyciete.shape[1],
        'width':  dane_przyciete.shape[2],
        'transform': transform_nowy,
        'nodata': -9999
    })

# Zapisz wynik
with rasterio.open('nmt_obszar.tif', 'w', **meta_nowe) as dst:
    dst.write(dane_przyciete)

print('Zapisano przyciety raster.')
```


````python
# [JUPYTER] Demonstracja przycinania rastra

try:
    import numpy as np
    import matplotlib.pyplot as plt
    import matplotlib.patches as mpatches
    from matplotlib.patches import Polygon as MplPolygon
    from matplotlib.collections import PatchCollection

    # Syntetyczny raster globalny
    H, W = 100, 150
    np.random.seed(7)
    dane_glob = np.random.randint(50, 1500, (H, W)).astype(float)

    # Obszar zainteresowania — prostokat (indeksy pikseli)
    r0, r1 = 20, 70   # wiersze
    c0, c1 = 40, 110  # kolumny

    # Przycinanie przez indeksowanie NumPy
    dane_przyciete = dane_glob[r0:r1, c0:c1]

    # Maska poza obszarem — piksele NoData
    dane_z_nodata = dane_glob.copy()
    dane_z_nodata[:r0, :] = np.nan
    dane_z_nodata[r1:, :] = np.nan
    dane_z_nodata[:, :c0] = np.nan
    dane_z_nodata[:, c1:] = np.nan

    fig, axes = plt.subplots(1, 3, figsize=(15, 4))

    # Oryginalny raster
    im0 = axes[0].imshow(dane_glob, cmap='terrain', aspect='auto')
    rect = mpatches.Rectangle((c0, r0), c1-c0, r1-r0,
                               linewidth=2, edgecolor='red', facecolor='none')
    axes[0].add_patch(rect)
    axes[0].set_title('Oryginalny raster\n(czerwony = obszar przycięcia)')
    plt.colorbar(im0, ax=axes[0])

    # Z NoData poza obszarem
    im1 = axes[1].imshow(dane_z_nodata, cmap='terrain', aspect='auto')
    axes[1].set_title('Po naloeniu maski\n(szary = NoData)')
    plt.colorbar(im1, ax=axes[1])

    # Przyciety raster
    im2 = axes[2].imshow(dane_przyciete, cmap='terrain', aspect='auto')
    axes[2].set_title(f'Przyciety raster\n({dane_przyciete.shape[1]}x{dane_przyciete.shape[0]} px)')
    plt.colorbar(im2, ax=axes[2])

    plt.suptitle('Clip by mask — przycinanie rastra do obszaru', fontsize=13)
    plt.tight_layout()
    plt.show()

    print(f'Oryginalny:   {dane_glob.shape[1]} x {dane_glob.shape[0]} px = {dane_glob.size:,} pikseli')
    print(f'Przyciety:    {dane_przyciete.shape[1]} x {dane_przyciete.shape[0]} px = {dane_przyciete.size:,} pikseli')
    print(f'Redukcja:     {(1 - dane_przyciete.size/dane_glob.size)*100:.1f}%')

except ImportError:
    print('Zainstaluj: pip install numpy matplotlib')
````

## 6. Reklasyfikacja i kalkulator rastrowy

### 6.1 Czym jest reklasyfikacja

Reklasyfikacja (ang. *reclassify*) zamienia wartosci ciagłe
na kategorie dyskretne. Przyklad:

```
Wartosc NMT    ->  Klasa
0   – 200 m        1  (niziny)
200 – 500 m        2  (wyzyny)
500 – 1500 m       3  (gory niskie)
> 1500 m           4  (gory wysokie)
```

### 6.2 Reklasyfikacja w PyQGIS przez Processing

```python
# [QGIS] Reklasyfikuj NMT na klasy wysokosciowe
import processing

raster = QgsProject.instance().mapLayersByName('NMT')[0]

# Tabela reklasyfikacji: [[min, max, nowa_wartosc], ...]
tabela = [
    [0,    200,  1],   # niziny
    [200,  500,  2],   # wyzyny
    [500,  1500, 3],   # gory niskie
    [1500, 9999, 4],   # gory wysokie
]

wynik = processing.run('native:reclassifybytable', {
    'INPUT_RASTER'    : raster,
    'RASTER_BAND'     : 1,
    'TABLE'           : [v for row in tabela for v in row],
    'NO_DATA'         : -9999,
    'RANGE_BOUNDARIES': 0,      # 0 = min <= x < max
    'NODATA_FOR_MISSING': True,
    'OUTPUT'          : 'TEMPORARY_OUTPUT'
})

reklasy = wynik['OUTPUT']
reklasy.setName('NMT_klasy')
QgsProject.instance().addMapLayer(reklasy)
print('Reklasyfikacja zakonczona.')
```

### 6.3 Reklasyfikacja przez NumPy (rasterio)

```python
import numpy as np

# dane = tablica NumPy z wartosciami NMT
```

### 6.4 Kalkulator rastrowy w PyQGIS

Kalkulator rastrowy pozwala pisac wyrazenia matematyczne
operujace na pasmach rastra.

```python
# [QGIS] Oblicz nachylenie terenu przez Processing
import processing

nmt = QgsProject.instance().mapLayersByName('NMT')[0]

# Nachylenie terenu
nachylenie = processing.run('gdal:slope', {
    'INPUT'          : nmt,
    'BAND'           : 1,
    'SCALE'          : 1,
    'AS_PERCENT'     : False,  # w stopniach
    'COMPUTE_EDGES'  : True,
    'OUTPUT'         : 'TEMPORARY_OUTPUT'
})

# Ekspozycja terenu (aspect)
ekspozycja = processing.run('gdal:aspect', {
    'INPUT'   : nmt,
    'BAND'    : 1,
    'OUTPUT'  : 'TEMPORARY_OUTPUT'
})

# Cieniowanie rzezby (hillshade)
hillshade = processing.run('gdal:hillshade', {
    'INPUT'       : nmt,
    'BAND'        : 1,
    'Z_FACTOR'    : 2,      # wyolbrzmienie pionowe
    'SCALE'       : 1,
    'AZIMUTH'     : 315,    # kierunek swiatla
    'ALTITUDE'    : 45,     # kat swiatla
    'OUTPUT'      : 'TEMPORARY_OUTPUT'
})

for wynik, nazwa in [
    (nachylenie, 'Nachylenie'),
    (ekspozycja, 'Ekspozycja'),
    (hillshade,  'Cieniowanie')
]:
    w = wynik['OUTPUT']
    w.setName(nazwa)
    QgsProject.instance().addMapLayer(w)

print('Wszystkie pochodne NMT dodane do projektu.')
```


````python
# [JUPYTER] Reklasyfikacja i pochodne rastra

try:
    import numpy as np
    import matplotlib.pyplot as plt
    import matplotlib.colors as mcolors

    np.random.seed(42)
    H, W = 100, 120
    x = np.linspace(0, 1, W)
    y = np.linspace(0, 1, H)
    xx, yy = np.meshgrid(x, y)

    nmt = (200
           + 1400 * np.exp(-((xx-0.3)**2 + (yy-0.4)**2) / 0.06)
           + 800  * np.exp(-((xx-0.7)**2 + (yy-0.6)**2) / 0.04)
           + 40   * np.random.randn(H, W))
    nmt = np.clip(nmt, 0, 2000)

    # Reklasyfikacja
    klasy = np.zeros_like(nmt, dtype=np.int8)
    klasy[(nmt >= 0)   & (nmt < 200)]  = 1  # niziny
    klasy[(nmt >= 200) & (nmt < 500)]  = 2  # wyzyny
    klasy[(nmt >= 500) & (nmt < 1000)] = 3  # gory niskie
    klasy[nmt >= 1000]                 = 4  # gory wysokie

    # Nachylenie (gradient)
    gy, gx = np.gradient(nmt)
    nachylenie = np.sqrt(gx**2 + gy**2)

    # Hillshade
    az_rad = np.radians(315)
    alt_rad = np.radians(45)
    hillshade = (np.cos(alt_rad) * np.cos(np.arctan(nachylenie))
                 + np.sin(alt_rad) * np.sin(np.arctan(nachylenie))
                 * np.cos(az_rad - np.arctan2(-gx, gy)))
    hillshade = np.clip(hillshade * 255, 0, 255)

    # Statystyki klas
    nazwy = {1: 'Niziny (<200m)', 2: 'Wyzyny (200-500m)',
             3: 'Gory niskie (500-1000m)', 4: 'Gory wysokie (>1000m)'}
    print('Reklasyfikacja NMT:')
    for k, n in nazwy.items():
        px = (klasy == k).sum()
        print(f'  Klasa {k} {n}: {px:5d} px ({px/klasy.size*100:.1f}%)')

    # Wizualizacja
    fig, axes = plt.subplots(2, 2, figsize=(13, 10))

    im0 = axes[0,0].imshow(nmt, cmap='terrain', aspect='auto')
    plt.colorbar(im0, ax=axes[0,0], label='m n.p.m.')
    axes[0,0].set_title('NMT — wartosci ciagłe')

    cmap_klasy = mcolors.ListedColormap(['#4FC3F7','#81C784','#A5D6A7','#D7CCC8'])
    im1 = axes[0,1].imshow(klasy, cmap=cmap_klasy, aspect='auto', vmin=0.5, vmax=4.5)
    cbar = plt.colorbar(im1, ax=axes[0,1], ticks=[1,2,3,4])
    cbar.ax.set_yticklabels(['Niziny', 'Wyzyny', 'Gory niskie', 'Gory wysokie'])
    axes[0,1].set_title('Reklasyfikacja — 4 klasy')

    im2 = axes[1,0].imshow(nachylenie, cmap='Reds', aspect='auto')
    plt.colorbar(im2, ax=axes[1,0], label='Gradient')
    axes[1,0].set_title('Nachylenie terenu (gradient NMT)')

    im3 = axes[1,1].imshow(hillshade, cmap='gray', aspect='auto')
    plt.colorbar(im3, ax=axes[1,1])
    axes[1,1].set_title('Cieniowanie rzezby (hillshade)')

    plt.suptitle('Reklasyfikacja i pochodne NMT', fontsize=13)
    plt.tight_layout()
    plt.show()

except ImportError:
    print('Zainstaluj: pip install numpy matplotlib')
````

## 7. Statystyki strefowe — zonal statistics

### 7.1 Co to jest

Statystyki strefowe (ang. *zonal statistics*) odpowiadaja na pytanie:
**jaka jest srednia (lub min, max, suma) wartosci rastra wewnatrz kazdego poligonu?**

Przyklady:

- Srednia wysokosc terenu per gmina (NMT + granice gmin)
- Srednie NDVI per pole uprawne (zobrazowanie + dzialki)
- Suma opadu per zlewnia (mapa opadu + granice zlewni)
- Maksymalna temperatura per park narodowy

### 7.2 Zonal statistics w PyQGIS przez Processing

```python
# [QGIS] Statystyki strefowe — NMT per gmina
import processing

nmt    = QgsProject.instance().mapLayersByName('NMT')[0]
gminy  = QgsProject.instance().mapLayersByName('granice_gmin')[0]

wynik = processing.run('native:zonalstatisticsfb', {
    'INPUT'        : gminy,    # warstwa wektorowa (strefy)
    'INPUT_RASTER' : nmt,      # raster
    'RASTER_BAND'  : 1,
    'COLUMN_PREFIX': 'nmt_',   # prefiks nowych pol atrybutow
    'STATISTICS'   : [0, 1, 2, 4, 5, 6],
    # 0=liczba, 1=suma, 2=srednia, 4=min, 5=max, 6=zakres
    'OUTPUT'       : 'TEMPORARY_OUTPUT'
})

gminy_ze_stat = wynik['OUTPUT']

# Sprawdz nowe pola atrybutow
print([p.name() for p in gminy_ze_stat.fields()])
# -> [..., 'nmt_count', 'nmt_sum', 'nmt_mean', 'nmt_min', 'nmt_max', 'nmt_range']

# Ranking gmin wg sredniej wysokosci
wyniki = []
for f in gminy_ze_stat.getFeatures():
    wyniki.append((f['NAZWA'], f['nmt_mean'], f['nmt_max']))

wyniki.sort(key=lambda x: x[1] or 0, reverse=True)
print('TOP 5 gmin wg sredniej wysokosci:')
for nazwa, srednia, maks in wyniki[:5]:
    print(f'  {nazwa:<30} srednia: {srednia:.0f}m  maks: {maks:.0f}m')
```

### 7.3 Dostepne statystyki

| Kod | Statystyka | Opis |
|---|---|---|
| 0 | Count | Liczba pikseli w strefie |
| 1 | Sum | Suma wartosci pikseli |
| 2 | Mean | Srednia wartosci |
| 3 | Median | Mediana |
| 4 | Min | Minimum |
| 5 | Max | Maksimum |
| 6 | Range | Roznica max - min |
| 7 | Minority | Najrzadsza wartosc (dla rastra kategorycznego) |
| 8 | Majority | Najczestsza wartosc |
| 9 | Variety | Liczba unikalnych wartosci |
| 10 | Variance | Wariancja |
| 11 | StDev | Odchylenie standardowe |

### 7.4 Zonal statistics przez rasterio + geopandas

```python
import rasterio
import geopandas as gpd
import numpy as np
from rasterio.mask import mask as rasterio_mask
from shapely.geometry import mapping

gminy  = gpd.read_file('gminy.gpkg')

wyniki = []

with rasterio.open('nmt.tif') as src:
    gminy_repr = gminy.to_crs(src.crs)   # ten sam CRS!

    for idx, row in gminy_repr.iterrows():
        try:
            dane_strefy, _ = rasterio_mask(
                src,
                [mapping(row.geometry)],
                crop=True,
                nodata=-9999
            )
            piksele = dane_strefy[0]          # pierwsze pasmo
            poprawne = piksele[piksele != -9999]

            if len(poprawne) > 0:
                wyniki.append({
                    'gmina'  : row['NAZWA'],
                    'n_px'   : len(poprawne),
                    'srednia': round(float(poprawne.mean()), 1),
                    'min'    : int(poprawne.min()),
                    'max'    : int(poprawne.max()),
                })
        except Exception:
            pass

import pandas as pd
df = pd.DataFrame(wyniki).sort_values('srednia', ascending=False)
print(df.head(10).to_string(index=False))
```


````python
# [JUPYTER] Demonstracja zonal statistics

try:
    import numpy as np
    import matplotlib.pyplot as plt
    import matplotlib.colors as mcolors
    import matplotlib.patches as mpatches

    np.random.seed(11)
    H, W = 100, 100

    # Syntetyczny raster (NMT)
    x = np.linspace(0, 1, W)
    y = np.linspace(0, 1, H)
    xx, yy = np.meshgrid(x, y)
    nmt = (300
           + 900 * np.exp(-((xx-0.25)**2 + (yy-0.3)**2) / 0.04)
           + 500 * np.exp(-((xx-0.75)**2 + (yy-0.7)**2) / 0.03)
           + 20  * np.random.randn(H, W))
    nmt = np.clip(nmt, 100, 1500)

    # Syntetyczne strefy (6 obszarow = 2x3 siatka)
    strefy = np.zeros((H, W), dtype=int)
    for r in range(2):
        for c in range(3):
            r0, r1 = r*50, (r+1)*50
            c0, c1 = c*33, (c+1)*33 if c < 2 else W
            strefy[r0:r1, c0:c1] = r*3 + c + 1

    # Oblicz statystyki per strefa
    stat = {}
    for s in range(1, 7):
        px = nmt[strefy == s]
        stat[s] = {
            'srednia': px.mean(),
            'min':     px.min(),
            'max':     px.max(),
            'std':     px.std(),
            'n':       len(px)
        }

    print(f'  {"Strefa":<10} {"Srednia":>9} {"Min":>7} {"Max":>7} {"Std":>7}')
    print('  ' + '-' * 48)
    for s, v in stat.items():
        print(f'  Strefa {s:<4} {v["srednia"]:>9.1f} {v["min"]:>7.1f} {v["max"]:>7.1f} {v["std"]:>7.1f}')

    # Mapa statystyk strefowych
    mapa_srednia = np.zeros((H, W))
    for s, v in stat.items():
        mapa_srednia[strefy == s] = v['srednia']

    fig, axes = plt.subplots(1, 3, figsize=(15, 4))

    im0 = axes[0].imshow(nmt, cmap='terrain', aspect='auto')
    plt.colorbar(im0, ax=axes[0], label='m n.p.m.')
    axes[0].set_title('Raster NMT')

    cmap_s = mcolors.ListedColormap(['#E3F2FD','#BBDEFB','#90CAF9','#64B5F6','#42A5F5','#1E88E5'])
    im1 = axes[1].imshow(strefy, cmap=cmap_s, aspect='auto', vmin=0.5, vmax=6.5)
    for s in range(1, 7):
        r = (s-1)//3 * 50 + 25
        c = ((s-1)%3) * 33 + 16
        axes[1].text(c, r, str(s), ha='center', va='center',
                     fontsize=14, fontweight='bold', color='white')
    axes[1].set_title('Strefy (obszary analizy)')

    im2 = axes[2].imshow(mapa_srednia, cmap='terrain', aspect='auto')
    plt.colorbar(im2, ax=axes[2], label='Srednia wys. (m)')
    for s in range(1, 7):
        r = (s-1)//3 * 50 + 25
        c = ((s-1)%3) * 33 + 16
        axes[2].text(c, r, f'{stat[s]["srednia"]:.0f}m',
                     ha='center', va='center', fontsize=9,
                     color='white', fontweight='bold')
    axes[2].set_title('Srednia wysokosc per strefa')

    plt.suptitle('Zonal Statistics — statystyki strefowe', fontsize=13)
    plt.tight_layout()
    plt.show()

except ImportError:
    print('Zainstaluj: pip install numpy matplotlib')
````

```python
import rasterio
import numpy as np

# Dane do zapisu
dane = np.array([[100, 200, 300],
```

````python
# [JUPYTER] Pełna wizualizacja wynikow analizy rastrowej

try:
    import numpy as np
    import matplotlib.pyplot as plt
    import matplotlib.colors as mcolors
    from matplotlib.gridspec import GridSpec

    np.random.seed(99)
    H, W = 120, 150
    x = np.linspace(0, 1, W)
    y = np.linspace(0, 1, H)
    xx, yy = np.meshgrid(x, y)

    nmt = (200
           + 1200 * np.exp(-((xx-0.3)**2 + (yy-0.35)**2) / 0.05)
           + 700  * np.exp(-((xx-0.72)**2 + (yy-0.65)**2) / 0.04)
           + 30   * np.random.randn(H, W))
    nmt = np.clip(nmt, 100, 1700)

    # Reklasyfikacja
    klasy = np.zeros_like(nmt, dtype=int)
    klasy[(nmt < 300)]                    = 1
    klasy[(nmt >= 300) & (nmt < 600)]     = 2
    klasy[(nmt >= 600) & (nmt < 1100)]    = 3
    klasy[nmt >= 1100]                    = 4

    # Nachylenie
    gy, gx = np.gradient(nmt)
    nachyl = np.sqrt(gx**2 + gy**2)

    # Hillshade
    hill = np.cos(np.radians(45)) * np.cos(np.arctan(nachyl))

    # Syntetyczne NDVI
    ndvi = np.clip(0.8 - 0.5 * (nmt / 1700)
                   + 0.15 * np.exp(-((xx-0.3)**2+(yy-0.35)**2)/0.1)
                   + 0.05 * np.random.randn(H, W), -0.2, 0.9)

    # Layout wykresu
    fig = plt.figure(figsize=(16, 10))
    gs = GridSpec(2, 3, figure=fig, hspace=0.35, wspace=0.3)

    # NMT
    ax0 = fig.add_subplot(gs[0, 0])
    im0 = ax0.imshow(nmt, cmap='terrain', aspect='auto')
    plt.colorbar(im0, ax=ax0, label='m n.p.m.', shrink=0.85)
    ax0.set_title('Numeryczny Model Terenu')
    ax0.axis('off')

    # Hillshade
    ax1 = fig.add_subplot(gs[0, 1])
    ax1.imshow(hill, cmap='gray', alpha=0.6, aspect='auto')
    im1 = ax1.imshow(nmt, cmap='terrain', alpha=0.6, aspect='auto')
    ax1.set_title('NMT z cieniowaniem rzezby')
    ax1.axis('off')

    # NDVI
    ax2 = fig.add_subplot(gs[0, 2])
    im2 = ax2.imshow(ndvi, cmap='RdYlGn', vmin=-0.2, vmax=0.9, aspect='auto')
    plt.colorbar(im2, ax=ax2, label='NDVI', shrink=0.85)
    ax2.set_title('NDVI (wskaznik roslinnosci)')
    ax2.axis('off')

    # Reklasyfikacja
    ax3 = fig.add_subplot(gs[1, 0])
    cmap_k = mcolors.ListedColormap(['#80DEEA','#A5D6A7','#BCAAA4','#ECEFF1'])
    norm_k = mcolors.BoundaryNorm([0.5,1.5,2.5,3.5,4.5], 4)
    im3 = ax3.imshow(klasy, cmap=cmap_k, norm=norm_k, aspect='auto')
    cbar3 = plt.colorbar(im3, ax=ax3, ticks=[1,2,3,4], shrink=0.85)
    cbar3.ax.set_yticklabels(['Niziny','Wyzyny','Gory niskie','Gory wys.'], fontsize=8)
    ax3.set_title('Reklasyfikacja (4 klasy)')
    ax3.axis('off')

    # Nachylenie
    ax4 = fig.add_subplot(gs[1, 1])
    im4 = ax4.imshow(nachyl, cmap='YlOrRd', aspect='auto')
    plt.colorbar(im4, ax=ax4, label='Gradient', shrink=0.85)
    ax4.set_title('Nachylenie terenu')
    ax4.axis('off')

    # Histogram NMT
    ax5 = fig.add_subplot(gs[1, 2])
    ax5.hist(nmt.ravel(), bins=50, color='steelblue', edgecolor='white', linewidth=0.4)
    ax5.axvline(300,  color='cyan',   linestyle='--', label='300m')
    ax5.axvline(600,  color='green',  linestyle='--', label='600m')
    ax5.axvline(1100, color='brown',  linestyle='--', label='1100m')
    ax5.set_title('Rozklad wysokosci (histogram)')
    ax5.set_xlabel('Wysokosc (m)')
    ax5.set_ylabel('Liczba pikseli')
    ax5.legend(fontsize=8)

    plt.suptitle('Kompleksowa analiza rastrowa — NMT i wskazniki pochodne',
                 fontsize=14, fontweight='bold')
    plt.show()

except ImportError:
    print('Zainstaluj: pip install numpy matplotlib')
````

## 9. PyQGIS vs rasterio — kiedy co uzywac

| Zadanie | PyQGIS | rasterio |
|---|---|---|
| Wczytanie rastra do projektu QGIS | Tak — `QgsRasterLayer` | Nie |
| Odczyt wartosci w punkcie | Tak — `dataProvider().sample()` | Tak — przez transformacje |
| Statystyki pasma | Tak — `bandStatistics()` | Tak — przez NumPy |
| Clip by mask | Tak — `gdal:cliprasterbymasklayer` | Tak — `rasterio.mask` |
| Reklasyfikacja | Tak — `native:reclassifybytable` | Tak — maski NumPy |
| Zonal statistics | Tak — `native:zonalstatisticsfb` | Tak — petla + mask |
| Obliczenia arytmetyczne (NDVI) | Tak — kalkulator rastrowy | Tak — NumPy (szybciej) |
| Hillshade, slope, aspect | Tak — `gdal:hillshade` itd. | Mozliwe — NumPy (trudniej) |
| Praca bez QGIS (serwer, Jupyter) | Nie | Tak |
| Integracja z data science | Slaba | Doskonala (pandas, sklearn) |
| Wizualizacja | QGIS Canvas | Matplotlib (wiecej kontroli) |

### Zasada

Jezeli pracujesz wewnatrz projektu QGIS i potrzebujesz
algorytmow Processing Toolbox — uzyj **PyQGIS**.

Jezeli robisz skomplikowane obliczenia numeryczne,
pracujesz w Jupyter lub na serwerze — uzyj **rasterio + NumPy**.

W praktyce czesto laczymy oba: PyQGIS do wczytania i clipowania,
rasterio do obliczen, PyQGIS ponownie do zapisu i wizualizacji.

## 10. Podsumowanie i slownik pojec

### Co omowilismy

- model danych rastrowych: piksel, pasmo, rozdzielczosc, NoData, geotransformacja,
- wczytywanie rastra przez `QgsRasterLayer` i `rasterio.open()`,
- odczyt metadanych: CRS, liczba pasm, rozdzielczosc, zasieg, statystyki,
- odczyt wartosci w punkcie przez `dataProvider().sample()` i `rowcol()`,
- operacje NumPy na pikselach: arytmetyka, maski logiczne, NDVI,
- przycinanie do obszaru: `gdal:cliprasterbymasklayer` i `rasterio.mask`,
- reklasyfikacja: `native:reclassifybytable` i maski NumPy,
- pochodne NMT: nachylenie, ekspozycja, hillshade przez Processing,
- statystyki strefowe: `native:zonalstatisticsfb` i petla rasterio,
- zapis rastra i wizualizacja w Matplotlib.

### Slownik pojec

| Pojecie | Definicja |
|---|---|
| **Piksel** | Podstawowa komorka rastra przechowujaca wartosc numeryczna |
| **Pasmo (band)** | Jedna warstwa wartosci w rastrze wielospektralnym |
| **Rozdzielczosc przestrzenna** | Rozmiar piksela na ziemi (np. 10 m × 10 m) |
| **Geotransformacja** | Szesc liczb laczacych numery wierszy/kolumn ze wspolrzednymi |
| **NoData** | Specjalna wartosc oznaczajaca brak danych (nie jest zerem!) |
| **GeoTIFF** | Format rastrowy przechowujacy geotransformacje i CRS wewnatrz pliku |
| **QgsRasterLayer** | Klasa PyQGIS reprezentujaca warstwe rastrowa |
| **rasterio** | Biblioteka Pythona do odczytu i zapisu rastrow, zwraca NumPy |
| **NumPy ndarray** | Wielowymiarowa tablica liczb — format danych rastrowych w Python |
| **NDVI** | Normalized Difference Vegetation Index — wskaznik zdrowia roslinnosci |
| **Clip by mask** | Przycinanie rastra do ksztaltu warstwy wektorowej |
| **Reklasyfikacja** | Zamiana wartosci ciaglych na kategorie dyskretne |
| **Hillshade** | Cieniowanie rzezby terenu — wizualizacja NMT |
| **Zonal statistics** | Obliczanie statystyk wartosci rastra wewnatrz stref (poligonow) |
| **NMT** | Numeryczny Model Terenu — raster wysokosci powierzchni gruntu |
| **bandStatistics()** | Metoda PyQGIS obliczajaca min, max, srednia, odch.std dla pasma |
